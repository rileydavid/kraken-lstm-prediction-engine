```python
import numpy as np 
import pandas as pd 
from datetime import datetime

from sklearn.linear_model import LinearRegression

```

```python
files = ['XBTUSD_full.csv', 'XBTUSD_Q1.csv', 'XBTUSD_Q2.csv']
df = pd.DataFrame()
for file in files:
    data = pd.read_csv(file, names=["time", "price", "volume"], header=None)
    df = pd.concat([df, data], axis=0)
# df.to_csv('XBTUSD.csv', index=False, header=None)

df.tail()
```

```python
# reduce size of dataframe
def split_dataframe(df, chunk_size = 5000000): 
    chunks = list()
    num_chunks = len(df) // chunk_size + 1
    for i in range(num_chunks):
        chunks.append(df[i*chunk_size:(i+1)*chunk_size])
    return chunks

df = split_dataframe(df)
```

```python
# Training data
X = df.loc[:, ['time']]  # features
y = df.loc[:, 'price']  # target

print(X)

# Train the model
model = LinearRegression()
model.fit(X, y)

# the training data
#y_pred = pd.Series(model.predict(X), index=X.index)
```
